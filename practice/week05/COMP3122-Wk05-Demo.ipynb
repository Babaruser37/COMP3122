## Block 1 — The Figure and the Axes  ·  Book §5.1

Two different objects: the Figure and the Axes. Clear this up now and the rest of the chapter
is easy; leave it fuzzy and every plot you write is guesswork.

**Figure** = the whole canvas, the thing you save.  **Axes** = one plot inside it, the thing
you draw on. *The Figure is the wall; an Axes is one painting hanging on it.*

```python
# explain: ask matplotlib for BOTH — a Figure, and one Axes inside it
fig, ax = plt.subplots()

ax.plot([1, 2, 3], [10, 20, 15])   # explain: draw on the AXES, not the figure
plt.show()                         # explain: in a script this opens the window
```

Check this in Python — ask it what the two objects actually are.

```python
fig, ax = plt.subplots()

print(type(fig))   # explain: matplotlib.figure.Figure
print(type(ax))    # explain: matplotlib.axes._axes.Axes  — a DIFFERENT class
plt.close(fig)     # explain: we are not drawing anything, so close it
```

### 🔴 LIVE DEMO — the error everyone hits first

**This cell raises on purpose.** Read the message: it is not saying *you called it wrong*, it
is saying *there is no such method*.

```python
fig, ax = plt.subplots()

fig.plot([1, 2, 3], [4, 5, 6])   # explain: RAISES — a Figure has no .plot method at all
```

**An empty plot appears under that traceback — that is expected.** `plt.subplots()` ran and
built the canvas successfully; only the *next* line failed. The blank picture is the Figure,
sitting there with nothing drawn on it. Which is the point.

The fix: replace `fig.plot` with `ax.plot`. Plotting happens on the **Axes**.

```python
fig, ax = plt.subplots()
ax.plot([1, 2, 3], [4, 5, 6])    # explain: this is the corrected line
plt.show()
```

### The minimum responsible plot

Data plus a title plus both axis labels. Anything less is a guess. The test: if you sent just
the picture to someone who has not seen your code, would they know what they were looking at?

```python
fig, ax = plt.subplots()
ax.plot([1, 2, 3], [10, 20, 15])
ax.set_title("A first labelled plot")    # explain: the headline above the plot
ax.set_xlabel("x value")                 # explain: text under the x-axis
ax.set_ylabel("y value")                 # explain: text beside the y-axis
plt.show()
```

> **Why capture `fig` when this example never uses it?** Because saving, resizing and spacing
> are all Figure operations: `fig.savefig("plot.png")`, `fig.set_size_inches(8, 5)`,
> `fig.tight_layout()`. Catching both halves costs four characters and saves rewriting the line
> later.

## Block 2 — Line plots and scatter plots  ·  Book §5.2

Both are dots on a grid. The difference is the line, and **the line is a claim**: it says
*this came after that*. Draw it where that is not true and the plot is a lie.

### 🔴 LIVE DEMO — one city's week

```python
cairo = weather.loc["Cairo"]        # explain: one row as a Series

fig, ax = plt.subplots()
ax.plot(cairo.index, cairo.to_numpy())   # explain: x = day names, y = temperatures
ax.set_title("Cairo daily highs")
ax.set_xlabel("Day of the week")
ax.set_ylabel("Temperature (°C)")
plt.show()
```

Thirty on Monday, up to thirty-four on Thursday, back to thirty by Sunday. You could have read
those seven numbers off the table — but you would not have *seen* the shape.

### Four cities on one Axes

Every `ax.plot` call **adds** a line. It does not replace the last one.

```python
fig, ax = plt.subplots()
for city in weather.index:                       # explain: one line per city
    ax.plot(weather.columns, weather.loc[city], label=city)

ax.set_title("Weekly highs by city")
ax.set_xlabel("Day of the week")
ax.set_ylabel("Temperature (°C)")
ax.set_ylim(16, 40)                              # explain: headroom, so the legend has room
ax.legend(loc="upper center", ncol=4)            # explain: four across, parked up there
plt.show()
```

> **Those last two lines are about where the legend goes.** Left alone, Matplotlib puts the box
> wherever it judges best — which, with four lines filling the frame, is usually on top of your
> data. *A legend that hides the data it is labelling is worse than no legend at all.*

Look at Toronto and Sydney: they **cross**. Sydney starts well above Toronto, drops below it on
Thursday, ties on Friday, then finishes above again. A table of twenty-eight numbers hides that
completely.

```python
# explain: check the crossing against the numbers rather than trusting the picture
print(weather.loc[["Toronto", "Sydney"]])
```

## Block 3 — Bar charts and histograms  ·  Book §5.3

Both draw rectangles standing on a baseline. The difference is what the x-axis **is**.

### A bar chart: one value per named section

```python
section_labels = enrollments["course"] + "-" + enrollments["section"]
print(section_labels.tolist())   # explain: ['COMP3122-A', 'COMP3122-B', ...]

fig, ax = plt.subplots()
ax.bar(section_labels, enrollments["enrolled"])   # explain: one bar per section
ax.set_title("Enrolment by section")
ax.set_xlabel("Course-section")
ax.set_ylabel("Students enrolled")
ax.tick_params(axis="x", rotation=30)   # explain: tilt — flat, six long names collide
plt.show()
```

> Try deleting the `tick_params` line and re-running. Six labels of the form `COMP3122-A` do

### A histogram: the shape of all 28 readings

Not about the cities any more — about the readings.

```python
all_temps = weather.to_numpy().flatten()   # explain: the 4x7 grid -> one flat array
print(all_temps.shape)                     # explain: (28,)

fig, ax = plt.subplots()
ax.hist(all_temps, bins=5, edgecolor="black")   # explain: 5 equal-width slices
ax.set_title("Distribution of all weekly highs (4 cities × 7 days)")
ax.set_xlabel("Temperature (°C)")
ax.set_ylabel("Number of readings")
plt.show()
```

Two clusters, not one — with **these five bins**. That qualifier matters: histogram shape
depends partly on how many bins you asked for. Do not take the picture on faith either — NumPy
will hand you the same numbers Matplotlib used to draw it.

```python
counts, edges = np.histogram(weather.to_numpy().flatten(), bins=5)

print("counts:", counts)          # explain: [7 3 9 4 5] — the five bars, left to right
print("edges :", edges.round(1))  # explain: 18.0 21.2 24.4 27.6 30.8 34.0

# explain: the data runs 18 to 34, a range of 16.  16 / 5 bins = 3.2 per bin
print("range :", all_temps.min(), "to", all_temps.max())
```

## Block 4 — Box plots  ·  Book §5.4

The one plot that answers *how do these groups compare?* in a single glance.

The **box** runs from the 25th percentile to the 75th — the middle 50% of the distribution.
The **line inside** is the median, not the mean. The **whiskers** reach out to the furthest
values that are not treated as outliers, and the **dots beyond them** are the outliers.

```python
# explain: matplotlib's boxplot does not speak DataFrame — it wants a LIST of arrays
se_scores = students.loc[students["program"] == "SE", "final_score"]
ds_scores = students.loc[students["program"] == "DS", "final_score"]
it_scores = students.loc[students["program"] == "IT", "final_score"]

fig, ax = plt.subplots()
ax.boxplot([se_scores, ds_scores, it_scores],   # explain: three arrays in a list
           tick_labels=["SE", "DS", "IT"])      # explain: names, in matching order
ax.set_title("Final score by program")
ax.set_ylabel("Final score")
plt.show()
```

> **`tick_labels=` is the Matplotlib 3.9 spelling.** Before 3.9 the same argument was called
> `labels=`, and you will still find `labels=` in older tutorials. If `tick_labels` raises a
> `TypeError` on your machine, your Matplotlib is older than 3.9.

Count the filtering steps: three filter lines before you get to the plot. Remember that number —
Block 5 does the identical picture in one line.

### The box is an interval, not a headcount

This is the part people get wrong. "Middle 50%" describes the **distribution**, not the number
of dots you can count inside the rectangle. Quartiles are interpolated *between* observations,
so with a handful of points the two come apart badly.

```python
# explain: check it on our own groups rather than believing the phrase
for name in ["SE", "DS", "IT"]:
    v = students.loc[students["program"] == name, "final_score"].sort_values()
    q1, q3 = np.percentile(v, [25, 75])
    inside = [x for x in v if q1 <= x <= q3]
    print(f"{name}: {list(v)}  Q1={q1}  Q3={q3}  ->  {len(inside)} of {len(v)} inside the box")
```

> SE and DS put **one of three** points inside the box. IT puts **none of two** inside it. The
> box is still telling you where the middle of the distribution is — it is not promising to
> enclose half the dots.
>
> Two more things a box plot cannot do. It cannot tell you a difference is *real* rather than
> chance — that is statistics, not a picture. And more data would **not** make the boxes wider:
> box width is a styling constant. What more data changes is the box's span along the value
> axis, which with three points is both unstable and biased small.

## Block 5 — Seaborn  ·  Book §5.5

Seaborn sits **on top of** Matplotlib. It does not replace it — everything you have learned
still applies, and we still hand it our own `ax`. What it replaces is the filtering steps.

```python
fig, ax = plt.subplots()
sns.boxplot(data=students, x="program", y="final_score", ax=ax)   # explain: ONE line
ax.set_title("Final score by program (Seaborn)")
ax.set_ylabel("Final score")   # explain: override seaborn's raw column name
plt.show()
```

Four lines of filtering became one call. `data=` takes the whole DataFrame, `x=` names the
grouping column, `y=` names the value column, and Seaborn does the splitting itself.

Note `ax=ax` on the end — that is us handing Seaborn *our* Axes. It is calling Matplotlib
underneath, so every `set_title` and `set_ylabel` you know still works.

### Seaborn labels your axes for you — from the raw column names

```python
fig, ax = plt.subplots()
sns.boxplot(data=students, x="program", y="final_score", ax=ax)

print(ax.get_xlabel())   # explain: 'program'
print(ax.get_ylabel())   # explain: 'final_score' — underscore and all
plt.close(fig)
```

> Helpful, until that underscore is projected in front of a client. Keep overriding it. And
> note what Seaborn does *not* do: it never auto-titles. It hands you the axis labels; the
> title is the one part of the minimum you still owe it every single time.

### `hue=` — one colour per category, legend included

```python
fig, ax = plt.subplots()
sns.scatterplot(data=students, x="hours_week", y="final_score",
                hue="program", ax=ax)   # explain: colour the dots by program
ax.set_title("Final score vs study hours, coloured by program")
plt.show()
```

In pure Matplotlib that would be three filters, three `ax.scatter` calls and a hand-built
legend — about seven lines instead of one.

### The pairplot — every pair of numeric columns at once

The single most useful plot for **starting** a project. Note there is no `plt.subplots()` here:
pairplot builds its own figure and grid.

```python
sns.pairplot(students[["program", "hours_week", "attendance", "final_score"]],
             hue="program")
plt.show()
```

> **⚠️ The pairplot scales badly.** N numeric columns gives an N×N grid — 25 plots at 5
> columns, 100 at 10, 900 at 30. On a wide DataFrame it produces a picture too small to read
> and can make the notebook appear frozen. Keep it to 3–6 hand-picked columns, as above.
> Choosing *which* columns is not a limitation — that choice is the analysis.
>
> The diagonal curves look strange on our data because two or three students per program is
> not enough to estimate a distribution. On real data they behave.

### The heatmap — a 2D table as coloured cells

`weather` is a **matrix layout**: its row and column labels are already the visual axes, so
you hand Seaborn the whole table.

```python
fig, ax = plt.subplots()
sns.heatmap(weather,
            annot=True,        # explain: print the number inside each cell
            fmt="d",           # explain: format it as a plain integer
            cmap="YlOrRd",     # explain: Yellow -> Orange -> Red
            ax=ax)
ax.set_title("Weekly highs grid (°C)")
plt.show()
```

> **Matrix vs long-form.** `weather` is a matrix — pass the whole table. `students` is
> long-form (one row per observation, one column per variable) — name the columns with `x=`
> and `y=`. Rule of thumb: if the table is already shaped like the plot, pass the table.
>
> **Colour maps.** *Sequential* (`YlOrRd`, `Blues`, `viridis`) for low→high in one direction.
> *Diverging* (`coolwarm`, `RdBu`) when the middle is meaningful — a correlation matrix, where
> zero is the neutral centre. *Categorical* (`tab10`, `Set1`) for unrelated groups.
>
> And a warning: the heatmap is the good plot most often used badly. It earns its place when
> the **cells** are the unit of meaning. For one column of numbers, a histogram says far more.

## Block 6 — Several plots on one figure  ·  Book §5.6

`plt.subplots(rows, cols)` returns a Figure and an **array** of Axes. How you index that array
depends on its shape — and the shape has one surprise in it.

```python
for r, c in [(2, 2), (1, 3), (3, 1)]:
    f, a = plt.subplots(r, c)
    print(f"plt.subplots({r}, {c})  ->  shape {a.shape}")
    plt.close(f)

# explain: (2,2) -> (2, 2)   2D, indexed axes[i, j]
# explain: (1,3) -> (3,)     FLAT
# explain: (3,1) -> (3,)     FLAT — and it looks nothing like (1,3) on the page
```

> **This is Matplotlib's doing, not NumPy's.** NumPy does not drop length-one dimensions —
> `np.zeros((1, 3)).shape` is still `(1, 3)`. What collapses the grid is `plt.subplots`'s
> default `squeeze=True`.

```python
print(np.zeros((1, 3)).shape)     # explain: (1, 3) — NumPy KEEPS the dimension

f, a = plt.subplots(1, 3, squeeze=False)
print(a.shape)                    # explain: (1, 3) — squeeze=False turns the collapsing off
plt.close(f)
```

## Block 7 — The habit, and the four traps  ·  Book §5.7–5.8

> **Always look at the data before you model it.**

It is easy to skip. The data is loaded, a model is waiting, and plotting first feels like a
detour. It is not. It is the step that catches the wrong file, the column you misread, the
units in the wrong scale, and the one outlier quietly moving every average you compute.

### ⚠️ The trap that does not raise

Common Mistakes #2. This cell **works**. It produces a plot, no warning, no error — and the
plot is wrong. Watch the number it prints.

```python
fig, ax = plt.subplots()
ax.plot(weather)   # explain: the WHOLE frame, no columns named — does NOT raise

print("lines drawn   :", len(ax.get_lines()))   # explain: SEVEN
print("weather.shape :", weather.shape)         # explain: (4, 7) — four cities, seven days
plt.show()
```

> You asked for four cities across seven days. You got **seven lines** — one per *column* —
> drawn across the four city names on the x-axis. Matplotlib draws one line per column, and
> nothing warned you.
>
> That is why *"be explicit about what is on each axis"* is the actual rule, and why an error
> you can see beats a silence you cannot.

### The other three traps

1. **Forgetting `plt.show()` in a script.** Notebooks render for you; a plain `.py` file does
   not. "My code runs but nothing happens" is almost always this.
2. **Picking a plot because of how it looks.** Heatmaps photograph well and pairplots look like
   effort. Start from the question, not the picture.
3. **No labels.** `set_title`, `set_xlabel`, `set_ylabel`, and `legend()` if there is more than
   one series. Four lines, and they answer *what am I looking at* forever.

### A plot can be truthful and still mislead

Nothing below is dishonest about any number. Run it and look at the two panels.

```python
regions = ["North", "South", "East", "West"]
sales   = [102, 104, 103, 106]

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].bar(regions, sales)
axes[0].set_title("y-axis starts at 0")
axes[0].set_ylabel("Units sold")

axes[1].bar(regions, sales)
axes[1].set_ylim(100, 107)          # explain: the whole plot is now 7 units tall
axes[1].set_title("y-axis starts at 100")
axes[1].set_ylabel("Units sold")

fig.tight_layout()
plt.show()

# explain: the real gap, in case the right-hand panel talked you into something
print(f"West vs North: {sales[3]} - {sales[0]} = {sales[3]-sales[0]} units "
      f"({(sales[3]/sales[0]-1)*100:.1f}%)")
print("bar heights above a y-axis starting at 100:", [v - 100 for v in sales])
```

> On the right, West's bar is **three times** the height of North's. The real difference is
> four units — about 4%.
>
> Is the right-hand panel *wrong*? Not necessarily. A truncated axis is the right choice when
> small differences genuinely matter — four units of body temperature is enormous. What tells
> the reader is the **tick labels**, the numbers up the side. `set_ylabel("Units sold")` names
> the quantity and says nothing about where the axis starts.
>
> **The dishonest chart is not the truncated axis. It is the truncated axis that does not
> admit it.**
