## Part 2 — Worked examples

One cell per section of Chapter 0. Read the section, run the cell, then change something and
run it again. Breaking an example on purpose is a good way to learn what it does.

### 0.1 — Values and types

````python
count   = 6              # explain: int - a whole number
average = 80.0           # explain: float - has a decimal point
name    = "Maya"         # explain: str - text, in quotes
passed  = True           # explain: bool - True or False

print(type(count), type(average), type(name), type(passed))

from_a_file = "88"       # explain: text, even though it looks numeric
print(from_a_file + "1") # 881 - text joined to text
print(int(from_a_file) + 1)   # 89 - converted first
````

### 0.2 — Lists

````python
scores = [88, 72, 95, 64, 79, 82]

print(len(scores), scores[0], scores[1], scores[-1])
print(scores[0:3])       # explain: 0, 1 and 2 - the stop is NOT included
print(scores[3:])        # explain: from 3 to the end

scores.append(91)        # explain: add to the end
scores[0] = 90           # explain: replace position 0
print(scores, 95 in scores)
````

### 0.3 — Tuples

````python
student = ("Maya", 88)
print(student[0], student[1], len(student))

# student[1] = 90        # explain: uncomment to see the TypeError

print(students[0])       # explain: the whole record for Maya [0]
print(students[0][1])    # explain: reach inside it for the score
````

### 0.4 — Dictionaries

````python
ages = {"Maya": 19, "Liam": 21, "Aisha": 20}

print(ages["Maya"], len(ages), "Liam" in ages)
print(ages.get("Diego"))          # explain: None instead of an error
print(ages.get("Diego", 0))       # explain: or a value you choose

for key in ages:                  # explain: looping gives the KEYS
    print(key, ages[key])

print(list(ages.items()))
````

### 0.5 — Conditions

````python
score = 88

if score >= 80:
    grade = "A"
elif score >= 70:                 # explain: only checked if the test above was False
    grade = "B"
else:
    grade = "C"
print(grade)

print(88 == 88, 88 != 72, 72 >= 72)
print(True and not False, 88 > 90 or 88 > 80)

if []:                            # explain: an empty list counts as False
    print("not empty")
else:
    print("empty things are False")
````

### 0.6 — Loops

````python
for student in students:
    print(student[0], student[1])     # explain: [0] name, [1] score

for i in range(3):
    print(i)                          # 0 1 2 - stops BEFORE 3

for score in [88, 72, 95]:
    if score < 80:
        continue                      # explain: skip the rest of this pass
    print("kept", score)

total = 0
i = 0
while total < 200 and i < len(students):   # explain: guard the index as well as the goal
    total = total + students[i][1]
    i = i + 1
print(total, "after", i, "students")
````

### 0.7 — Functions

````python
def add(a, b):
    return a + b                      # explain: hand the answer back

def passed(score, pass_mark=70):      # explain: default used unless you override it
    return score >= pass_mark

def average_score(records):
    total = 0
    for record in records:
        total = total + record[1]     # explain: [1] is the score
    return total / len(records)

print(add(3, 4), passed(88), passed(65, 60), average_score(students))
````

### 0.8 — Strings

````python
name = "  Maya  "
print(repr(name.strip()), repr(name.upper()), repr(name))   # explain: original unchanged

name = name.strip()          # explain: to KEEP the result, assign it back
print(repr(name))            # 'Maya'    - now the name itself is clean

line = "Maya,88"
parts = line.split(",")               # explain: cut wherever a comma appears
print(parts, int(parts[1]))

print("-".join(["a", "b", "c"]))
print("Maya scored 88".replace("88", "90"))
````

### 0.9 — Files

````python
# This cell writes its own small file first, so nothing is missing.
with open("ch0_demo.txt", "w") as f:      # explain: "w" ERASES the file first
    f.write("Maya,88\n")
    f.write("Liam,72\n")

with open("ch0_demo.txt") as f:          # explain: closed automatically at the end
    for line in f:
        clean = line.strip()             # explain: remove the newline
        if clean:
            parts = clean.split(",")
            print(parts[0], int(parts[1]))

import os; os.remove("ch0_demo.txt")     # explain: tidy up after the demo
````

### 0.10 — Reading an error message

````python
ages = {"Maya": 19, "Liam": 21, "Aisha": 20}

try:
    print(ages["Diego"])
except KeyError as e:
    print("Error type :", type(e).__name__)
    print("Message    :", e)
    print()
    print("Read the LAST line of a traceback first. It names the problem.")
    print("The line above it shows where in your code it happened.")
````
