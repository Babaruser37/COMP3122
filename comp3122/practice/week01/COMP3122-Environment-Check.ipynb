## 1. Python itself

```python
import sys, platform

print("Python version :", sys.version.split()[0])
print("Runs on        :", platform.system(), platform.release())

major, minor = sys.version_info[:2]
python_ok = (major == 3 and minor >= 9)
print()
print("OK      Python 3.9 or newer" if python_ok else
      f"PROBLEM Python {major}.{minor} is older than this course expects (3.9+)")
```

## 2. The libraries this course uses

Chapter 1 needs none of these — but Week 2 onward does, so it is worth knowing now
rather than in the middle of a graded lab.

```python
import importlib

libraries = ["numpy", "pandas", "matplotlib", "seaborn", "sklearn"]
found, missing = {}, []

for name in libraries:
    try:
        module = importlib.import_module(name)
        found[name] = getattr(module, "__version__", "unknown")
    except ImportError:
        missing.append(name)

for name, version in found.items():
    print(f"OK      {name:12} {version}")

for name in missing:
    print(f"PROBLEM {name:12} not installed")

libraries_ok = (len(missing) == 0)
```

## 3. Can you actually use them?

```python
import numpy as np
import pandas as pd

temps = np.array([[8, 10, 12], [30, 31, 29]])
table = pd.DataFrame(temps, index=["Toronto", "Cairo"], columns=["Mon", "Tue", "Wed"])

print(table)
print()
print("Mean per city:")
print(table.mean(axis=1))

use_ok = (table.shape == (2, 3))
print()
print("OK      NumPy and Pandas both work" if use_ok else "PROBLEM the table came out the wrong shape")
```

## 4. Can you draw a plot?

```python
import matplotlib
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(4, 2.2))
ax.plot([0, 1, 2, 3], [1, 4, 2, 5], marker="o")
ax.set_title("If you can see this line, plotting works")
ax.set_xlabel("x")
ax.set_ylabel("y")
plt.show()

plot_ok = True
print("OK      a plot appeared above" )
print("        (if you see no picture, that is a PROBLEM - tell your professor)")
```

## 5. Can you write and read a file?

```python
from pathlib import Path

test_file = Path("comp3122_write_test.txt")

try:
    test_file.write_text("hello from COMP3122\n")
    contents = test_file.read_text().strip()
    test_file.unlink()                 # tidy up after ourselves
    file_ok = (contents == "hello from COMP3122")
except Exception as error:
    file_ok = False
    print("Error:", error)

print("OK      you can write and read files" if file_ok else
      "PROBLEM writing a file failed - you may not have permission in this folder")
```

## 6. Summary

```python
checks = {
    "Python 3.9 or newer" : python_ok,
    "Libraries installed" : libraries_ok,
    "NumPy and Pandas run": use_ok,
    "Plotting works"      : plot_ok,
    "Files can be written": file_ok,
}

print("=" * 46)
for label, passed in checks.items():
    print(f"{'OK     ' if passed else 'PROBLEM'}  {label}")
print("=" * 46)

if all(checks.values()):
    print()
    print("Everything works. You are ready for Week 2.")
else:
    print()
    print("Something above needs fixing. Bring this notebook to your professor THIS WEEK.")
```
