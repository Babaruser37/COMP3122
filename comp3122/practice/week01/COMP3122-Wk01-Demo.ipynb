## Block 3 — Comprehensions and Lambdas  ·  Book §§1.4–1.5

### Before: you write every step

````python
# explain: the long way -- start empty, append one name at a time
names = []
for name, score in students:
    names.append(name)

print(names)
````

### After: you describe the result

````python
names = [name for (name, score) in students]   # explain: the same five lines, as one

print(names)
print(students)      # explain: unchanged - a comprehension returns a NEW list
````

### One pattern, three uses

````python
print([name  for (name, score) in students])                   # explain: every name
print([name  for (name, score) in students if score >= 85])    # explain: the if filters
print([score for (name, score) in students if score >= 70])    # explain: take the score instead
````

### From `def` to `lambda`

The two cells below do exactly the same work.

````python
def get_score(s):        # explain: a normal, named function
    return s[1]

print(sorted(students, key=get_score))   # explain: key says WHICH part to sort on
````

````python
print(sorted(students, key=lambda s: s[1]))                    # explain: a lambda does the same job inline -- lowest first
print(sorted(students, key=lambda s: s[1], reverse=True)[:3])  # explain: reverse then slice -- top three
````

## Block 4 — Arguments and Sequence Helpers  ·  Book §§1.6–1.7

### `*args` and `**kwargs`

````python
def total(*args):              # explain: collects however many arrive -- args is a tuple
    return sum(args)

print(total(88, 72, 95))       # explain: three here, but any number works

def describe(**kwargs):        # explain: collects NAMED arguments -- kwargs is a dict
    for key, value in kwargs.items():
        print(f"{key}: {value}")

describe(name="Maya", score=88)   # explain: each name= becomes a key
````

### The four built-in helpers

````python
names  = [name  for (name, score) in students]
scores = [score for (name, score) in students]

print(list(map(str.upper, names)))              # explain: map applies a function to every item
print(list(filter(lambda s: s >= 70, scores)))  # explain: filter keeps the ones that pass
print(list(zip(names, scores))[:2])             # explain: zip pairs two lists together
print(list(enumerate(names))[:2])               # explain: enumerate adds the position
````

### The trap: an iterator is walked once

Watch the second `list(pairs)` come back empty.

````python
a = [1, 2, 3]
b = ['x', 'y', 'z']

pairs = zip(a, b)         # explain: zip returns an ITERATOR, not a list
print(list(pairs))        # explain: full result
print(list(pairs))        # explain: exhausted after one pass -- []  - the same object, now empty

print(list(zip(a, b)))    # explain: full result again - this is a NEW zip
````

## Block 5 — Unpacking, f-strings, Generators  ·  Book §§1.8–1.9

### Tuple unpacking

````python
pair = ("Aisha", 95)
name, score = pair    # explain: unpacking -- two names left, one tuple right
print(name)
print(score)

for name, score in students:   # explain: unpacking in the for line -- no students[0] anywhere
    print(name, score)
````

*The next cell raises `ValueError` on purpose. That is the message to recognise.*

````python
try:
    a, b = ("x", "y", "z")     # explain: two names, three values -- deliberately wrong
except ValueError as e:
    print("ValueError:", e)    # explain: the counts have to match exactly
````

### f-strings

````python
name, score = ("Aisha", 95)

print(f"{name} scored {score}")   # explain: the f prefix lets {} hold real values
print(f"{score / 100:.2f}")       # explain: :.2f rounds to two decimals
print(f"{name:8}|")               # explain: :8 pads to eight characters -- watch the |

for i, (name, score) in enumerate(students):   # explain: enumerate plus unpacking, one line
    print(f"{i}: {name:8} {score:3}")          # explain: padding both columns lines the table up
````

### Generators

````python
def passing_scores(students):    # explain: yield instead of return makes this a generator
    for name, score in students:
        if score >= 70:
            yield score          # explain: hand back one value, then pause

for s in passing_scores(students):   # explain: the loop is what drives it
    print(s)
````

````python
passing = (score for (name, score) in students if score >= 70)   # explain: round brackets, not square
squares = (x * x for x in [1, 2, 3])

print(passing)              # explain: a generator object, nothing computed yet
print(list(passing))        # explain: now it runs
print(list(squares))        # explain: same idea, different expression
````

Same answer, less code, fewer places to make a mistake.
