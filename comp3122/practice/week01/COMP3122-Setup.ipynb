# COMP3122 — Setup and weekly fetch

This notebook does two things: it creates your `comp3122/` course folder (if it isn't there already), and it
downloads each week's notebooks from the course website into the right place inside it.

**How to use it:** run the first code cell once, then run the cell for the week you need. Re-running a week is
safe — nothing you have edited is ever overwritten.

```python
# Run this cell first (Shift+Enter). It defines the helpers used below — it changes nothing by itself.
import io, re, shutil, urllib.request, zipfile
from pathlib import Path

SITE = "https://comp3122.gblearn.com/notebooks"
ROOT = Path.home() / "comp3122"
TREE = ["practice", "labs", "assignments/a1", "assignments/a2", "assignments/project"]
LAB_RE = re.compile(r"COMP3122-Lab(\d\d)-Student\.ipynb$")

def make_tree():
    for d in TREE:
        (ROOT / d).mkdir(parents=True, exist_ok=True)
    print("Course folder:", ROOT)

def place(name, week):
    m = LAB_RE.search(name)
    if m and int(m.group(1)) >= 2:
        return ROOT / "labs" / f"lab{int(m.group(1)):02d}" / name
    return ROOT / "practice" / f"week{week:02d}" / name

def fetch_week(week):
    make_tree()
    url = f"{SITE}/comp3122-week{week:02d}.zip"
    try:
        data = urllib.request.urlopen(url, timeout=60).read()
    except Exception as e:
        print(f"Week {week:02d}: could not download {url} — {e}"); return
    added, kept = [], []
    with zipfile.ZipFile(io.BytesIO(data)) as z:
        for info in z.infolist():
            if info.is_dir() or "__MACOSX" in info.filename or ".ipynb_checkpoints" in info.filename:
                continue
            dest = place(Path(info.filename).name, week)
            dest.parent.mkdir(parents=True, exist_ok=True)
            if dest.exists():
                kept.append(dest); continue
            with z.open(info) as src, open(dest, "wb") as out:
                shutil.copyfileobj(src, out)
            added.append(dest)
    for p in added: print("  added ", p.relative_to(ROOT))
    for p in kept:  print("  kept  ", p.relative_to(ROOT), "(already there — not overwritten)")
    print(f"Week {week:02d}: {len(added)} new file(s), {len(kept)} already present.")

def status():
    make_tree()
    for p in sorted(ROOT.rglob("*")):
        if p.is_file() and ".ipynb_checkpoints" not in str(p):
            print(p.relative_to(ROOT))

print("Ready. Now run the cell for the week you want.")
```
