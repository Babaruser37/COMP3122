## Part 6 — Replace the loop

This is the task the week was for.

**Task 6.1.** Here is working Python. It adds 13% tax to seven prices and rounds each
result to two decimals.

```python
prices = [12.50, 8.30, 5.75, 19.99, 3.20, 14.40, 7.85]
tax_rate = 0.13

prices_with_tax = []
for p in prices:
    prices_with_tax.append(round(p * (1 + tax_rate), 2))
```

Rewrite it with NumPy. **No loop, no comprehension.** Same numbers out.

````python
prices = np.array([12.50, 8.30, 5.75, 19.99, 3.20, 14.40, 7.85])
tax_rate = 0.13

t61 = np.round(prices * (1 + tax_rate), 2)
````

````python
expected = [round(p * 1.13, 2) for p in [12.50, 8.30, 5.75, 19.99, 3.20, 14.40, 7.85]]
check('6.1 vectorized tax', t61, expected)
````
